```python
import plotly.graph_objects as go


def update_plot_layout(fig):
    """
    Updates the layout of the Plotly figure with consistent styling for axis titles, legends,
    dimensions, and other non-data-related visual elements.

    Args:
    fig (plotly.graph_objects.Figure): The figure to update.
    """
    fig.update_layout(
        template="plotly_white",
        autosize=True,
        xaxis=dict(
            title_font=dict(size=25, family="Arial", color="black"),
            tickfont=dict(size=14),
            gridcolor="lightgrey",
        ),
        yaxis=dict(
            title_font=dict(size=25, family="Arial", color="black"),
            tickfont=dict(size=14),
            gridcolor="lightgrey",
        ),
        margin=dict(l=50, r=50, t=50, b=50),
        width=800,
        height=500,
    )

    return fig
```

```python
import plotly.graph_objs as go
import statsmodels.api as sm
from clock_project.genome_analysis.analytical_correlation_analysis import (
    remove_outliers_iqr,
)


def plot_scatter_for_time(df, x_col, y_col, time):
    # Filter the dataframe for the specified time point
    sub_df = df[df["Time"] == time]
    x_data = sub_df[x_col]
    y_data = sub_df[y_col]

    # Remove outliers (assuming you have the remove_outliers_iqr function)
    x_data_no_outliers, y_data_no_outliers = remove_outliers_iqr(x_data, y_data)

    # Create the figure
    fig = go.Figure()

    # Scatter plot for the actual data points
    fig.add_trace(
        go.Scatter(
            x=x_data_no_outliers,
            y=y_data_no_outliers,
            mode="markers",
            name=f"Time {time}",  # Legend entry with the time label
            marker=dict(
                color="#62be9f",
                size=3,  # Adjust marker size as needed
            ),
        )
    )

    # Calculate OLS trendline
    x = sm.add_constant(x_data_no_outliers)  # adding a constant for OLS
    model = sm.OLS(y_data_no_outliers, x).fit()
    trendline = model.predict(x)

    # Add the trendline
    fig.add_trace(
        go.Scatter(
            x=x_data_no_outliers,
            y=trendline,
            mode="lines",
            line=dict(color="#c73d47", width=2),  # Deep red trendline
            showlegend=False,  # Hide legend for the trendline
        )
    )

    return fig


fig_exmple = plot_scatter_for_time(
    df_high_5000, "JSD_difference", "ENS_abs_difference", 0.5
)
# Annotate R² value
fig_exmple.add_annotation(
    xref="x domain",
    yref="y domain",
    x=0.95,
    y=0.95,
    text=f"\u03c1 = {0.43}",
    showarrow=False,
    font=dict(size=15, color="red"),
    align="right",
    ax=0,
    ay=0,
    bordercolor="black",
    borderwidth=1,
    borderpad=4,
    bgcolor="white",
    opacity=0.8,
)
# Consolidated update for axis titles
fig_exmple.update_xaxes(title_text=r"$\delta (\text{JAD})$")
fig_exmple.update_yaxes(title_text=r"$\delta (\text{ENS})$")
fig_exmple = update_plot_layout(fig_exmple)

fig_exmple.update_layout(
    showlegend=False,
    height=350,
    width=800,
    margin=dict(l=50, r=10, t=30, b=10),
)

fig_exmple.show()
fig_exmple.write_image("simulation_scatter_example_jad_simulation_5000.pdf")
```

```python
import pandas as pd
import plotly.graph_objects as go

# Data provided
x_values = [0.5, 1.0, 1.5, 2.0]

correlation_results_high_jad_5000 = {0.5: 0.44, 1.0: 0.46, 1.5: 0.36, 2.0: 0.34}
correlation_results_low_jad_5000 = {0.5: 0.37, 1.0: 0.26, 1.5: 0.18, 2.0: 0.16}

# Combine all data into a dataframe
data = {
    "Time": x_values * 2,
    "Measure": ["JAD"] * 4 + ["JAD"] * 4,
    "High/Low": ["Balanced"] * 4 + ["Imbalanced"] * 4,
    "Correlation": list(correlation_results_low_jad_5000.values())
    + list(correlation_results_high_jad_5000.values()),
}

# Create DataFrame
df_correlation = pd.DataFrame(data)

# Create a new figure
fig = go.Figure()

# Define color mapping for each measure
color_map = {"JAD": "#62be9f"}

# Add traces for each measure and condition (Balanced/Imbalanced)
for measure in df_correlation["Measure"].unique():
    for condition in ["Balanced", "Imbalanced"]:
        filtered_data = df_correlation[
            (df_correlation["Measure"] == measure)
            & (df_correlation["High/Low"] == condition)
        ]
        fig.add_trace(
            go.Scatter(
                x=filtered_data["Time"],
                y=filtered_data["Correlation"],
                mode="lines+markers",
                name=f"{condition}",
                line=dict(
                    dash="dash" if condition == "Balanced" else "solid",
                    color=color_map[measure],
                ),
                marker=dict(symbol="circle" if condition == "Balanced" else "square"),
            )
        )

fig = update_plot_layout(fig)

# Update layout
fig.update_layout(
    title=None,
    xaxis_title="<b>Time</b>",
    yaxis_title=r"$\hat{\rho}$",
    xaxis=dict(title_font=dict(size=25.5, color="black")),
    legend_title="Sequence Composition",
    showlegend=True,
    height=350,
    width=800,
    margin=dict(l=50, r=50, t=10, b=5),
    legend=dict(
        orientation="h",  # Horizontal legend
        yanchor="bottom",
        y=-0.3,  # Position legend below the plot
        xanchor="center",
        x=0.5,
        font=dict(size=14),
    ),
)
# Show the plot
fig.show()
fig.write_image("correlation_factor_time_difference_jad_simulation_500.pdf")
```

```python
from scipy.stats import spearmanr


# Function to compute Spearman correlation for any two columns
def compute_spearman(group, column1, column2):
    return spearmanr(group[column1], group[column2]).correlation


# Group by the 'Time' column and compute Spearman correlation for modifiable columns
def spearman_by_time(dataframe, column1, column2):
    return dataframe.groupby("Time").apply(
        lambda group: compute_spearman(group, column1, column2)
    )


# Example usage with modifiable column names
result = spearman_by_time(df_high_5000, "Ingroup_JSD", "ENS_abs_difference")

# Display the result
print(result)
```
